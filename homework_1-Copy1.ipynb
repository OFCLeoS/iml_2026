#### Data Visualization using PCA

```python
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

Firstly, lets project the 3D points to 2D using PCA.

```python
pca = PCA(n_components=2)
ulu_pca = pca.fit_transform(ulu)

plt.scatter(ulu_pca[:,0], ulu_pca[:,1])
plt.show()
```

The projection creates a densily packed rectangle in 2D. This does not give a lot of meaningful information on how the dataset was produced (which is what was requested).

On the graph, we have a visual on the data using the first and second PC as the new axis references. Since we do not have enough information, we will analyze the third PC, which is orthogonal to both the first and second PC.

To perform this analysis, we will use a histogram.

```python
pca = PCA(n_components=3)
ulu_pca = pca.fit_transform(ulu)

plt.hist(ulu_pca[:,2])
plt.show()
```

This histogram suggests that there are 2 significant "clusters" of a the z-coordinates in the data. To "visualize" this in our heads, we can think that a lot (most even) of the points are spread around the same z-axis value, and the rest are "pushed backwards".

We can try visualizing this using a scatter plot where the colours of the points are decided by their z-axis value. 

```python
plt.scatter(ulu_pca[:,0], ulu_pca[:,1], c=ulu_pca[:,2])
plt.show()
```

We can now clearly see how this dataset was produced. It was made by having data points that spell out "uni.lu" be pushed back on the z-axis.

#### K-Means

Since a k was not specified, we will use k = 2. Our reasoning is that the data points are spread within 2 clusters: The plane containing the "uni.lu" text that was "carved out", and the actual "uni.lu" text.

We will still use the elbow method to verify that this is a good choice.

```python
wcss_list = []
km_list = []
for k in range(1, 15):
    km = KMeans(n_clusters=k)
    km.fit(ulu)
    km_list.append(km)
    wcss_list.append(km.inertia_)

plt.plot(range(1, 15), wcss_list, marker="o")
plt.xlabel("k")
plt.ylabel("WCSS")
plt.show()
```

Upon visual inspection of the graph, k = 2 is indeed a good choice, although k = 3 is arguably better if we strictly follow the elbow method, which we won't, because we have a clear seperation between the 2 clusters upon visually inspection of the data.

To visualize this, we will project the data in 2D using the previous PCA values, and apply their colour depending on what cluster they belong to.

```python
kmeans = km_list[1]

plt.scatter(ulu_pca[:,0], ulu_pca[:,1],c=kmeans.labels_)
plt.show()
```

We can see that k-means clustering is an awful method to choose from. We believe that this is due to how it works, which is using distances, which, in this case, is not prefered.

#### DBSCAN

Since no parameters were specified, we will be using the default ones. Upon some research online, it seems that Standardizing the data before using DBSCAN is essential, so we will be doing that.

```python
# Standardizing is essential for DBSCAN when using default params
ulu_scaled = StandardScaler().fit_transform(ulu)

dbscan = DBSCAN().fit(ulu_scaled)

plt.scatter(ulu_pca[:,0], ulu_pca[:,1],c=dbscan.labels_)
plt.show()
```
